## Test 2: Read one order from MySQL

```python
from app.resources.OrderResource import OrderResource

order_resource = OrderResource()

order = order_resource.get_by_id("10101")
order
```

## Test 3: Read all order details for one order

```python
from app.resources.OrderDetailsResource import OrderDetailsResource

order_details_resource = OrderDetailsResource()

order_details = order_details_resource.get_by_id("10101")
order_details
```

## Test 4: Read one exact order detail using the composite key

```python
exact_order_detail = order_details_resource.get_by_order_and_product(
    "10101",
    "S18_2325"
)

exact_order_detail
```

## Test 5: Update one exact order detail row using the composite key

```python
from app.resources.OrderDetailsResource import OrderDetail

updated_detail = OrderDetail(
    orderNumber=10101,
    productCode="S18_2325",
    quantityOrdered=26,
    priceEach="108.06",
    orderLineNumber=4,
)

updated_count = order_details_resource.put_by_order_and_product(
    "10101",
    "S18_2325",
    updated_detail,
)

print("Rows updated:", updated_count)

order_details_resource.get_by_order_and_product("10101", "S18_2325")
```

## Test 6: Restore the original order detail value

```python
restored_detail = OrderDetail(
    orderNumber=10101,
    productCode="S18_2325",
    quantityOrdered=25,
    priceEach="108.06",
    orderLineNumber=4,
)

restored_count = order_details_resource.put_by_order_and_product(
    "10101",
    "S18_2325",
    restored_detail,
)

print("Rows restored:", restored_count)

order_details_resource.get_by_order_and_product("10101", "S18_2325")
```
